### Manage secondary occupation

We will have to code a part of the categories because their number is too small, which makes the variable too sparse.

```python
### Group and count: secondary occupation
# We observe some dispersion that requires grouping the categories of the variable
df_count = df_pm.groupby('occupation_sec1').size()
df_count = pd.DataFrame(df_count.sort_values(ascending = False))
df_count.columns=['number']
print(len(df_count))
print(df_count.iloc[:50])

```

```python
# Since there's a lot of “noise,” it's necessary to group the job categories together so as not to skew the graphs.
def categoriser_occupation_sec1(occupation_sec1):
    occupation_sec1 = str(occupation_sec1).lower()
    if 'photographer' in occupation_sec1 \
        or 'photojournalist' in occupation_sec1 \
        or 'war-photographer' in occupation_sec1 \
        or 'camera-operator' in occupation_sec1 \
        or 'cinematographer' in occupation_sec1 \
        or 'painter' in occupation_sec1 \
        or 'illustrator' in occupation_sec1 \
        or 'filmmaker' in occupation_sec1:
        output='visual_media'    
    elif 'journalist' in occupation_sec1 \
        or 'reporter' in occupation_sec1 \
        or 'television-journalist' in occupation_sec1 \
        or 'announcer' in occupation_sec1 \
        or 'radio-journalist' in occupation_sec1:
        output='journalism_core'
    elif 'writer' in occupation_sec1 \
        or 'novelist' in occupation_sec1 \
        or 'memoirist' in occupation_sec1 \
        or 'translator' in occupation_sec1 \
        or 'historian' in occupation_sec1:
        output='writing_intellectual'   
    elif 'nan' in occupation_sec1:
        output='other'
    else:
        output='other'
    return output  
```

```python
df_pm['occupation_sec1_grouped'] = df_pm['occupation_sec1'].apply(categoriser_occupation_sec1)
```

```python
df_count = df_pm.groupby('occupation_sec1_grouped').size()
df_count = pd.DataFrame(df_count.sort_values(ascending = False))
df_count.columns = ['number']
```

```python
print(df_count)
```

```python
## We only take the first 19 elements in the list, the more interesting ones
ldf = df_count[df_count.number > 5].index.to_list()
print(ldf)
```

```python
## prepare the function to code the data
def code_occupation_sec1(occup, ldf):

    """
    codes values according to list
    """

    if occup in ldf:
        output = occup
    else:
        output= 'other'
    return output        
```

```python
### Code the data
df_pm.occupation_sec1 = df_pm.occupation_sec1.apply(lambda x: code_occupation_sec1(x, ldf))
```

### Complete disjunctive table 

Sometimes also referred to as a dummy variable matrix or indicator matrix.

In the context of Multiple Correspondence Analysis (MCA), this table is used to transform qualitative (categorical) variables into a quantitative format suitable for analysis. Each category of a qualitative variable becomes a separate binary column (0 or 1), indicating the absence or presence of that category for each individual.

```python
## Complete disjunctive table
X = pd.get_dummies(DActives,prefix='',prefix_sep='')*1
X.head(3)
```

```python
### Nombre total de modalités, toute variable confondue
M = X.shape[1]
print('Number of categories:', M)
#nombre max de facteurs
Hmax = M-p
print('Maximum number of factors:', Hmax)
```

```python
Xm = X.copy(deep=True)
#Total sum per row: 
Xm.loc[:,'Total'] = Xm.sum(axis=1)
#Total sum per column: 
Xm.loc['Total',:] = Xm.sum(axis=0)
Xm.tail()
```

```python
import pandas as pd
import numpy as np
from fanalysis.mca import MCA

# 1. Selection and Basic Cleaning
# ------------------------------
cols = ['gender', 'coded_country', 'occupation_sec1', 'coded_employer']
# On crée une copie propre pour éviter les warnings
DActives = df_pm[cols].copy()

# On vire les vrais NaN et on s'assure que tout est du texte (string)
DActives = DActives.dropna().astype(str)

# 2. Security Check: Remove Constant Columns
# -----------------------------------------
# Une colonne avec 1 seule valeur (ex: que des 'other') fait exploser l'ACM
cols_to_keep = [c for c in DActives.columns if DActives[c].nunique() > 1]
DActives = DActives[cols_to_keep]

# 3. Security Check: Remove Dead Categories
# -----------------------------------------
# On s'assure qu'aucune ligne ou colonne ne finit à zéro après encodage
# (Certains espaces vides ou strings bizarres peuvent tromper fanalysis)
DActives = DActives.replace(r'^\s*$', 'unknown', regex=True) # Remplace les textes vides par 'unknown'

print(f"### Final Data Status ###")
print(f"Rows: {DActives.shape[0]}")
print(f"Active Variables: {DActives.columns.tolist()}")

# 4. Fit MCA with Error Handling
# ------------------------------
try:
    acm = MCA(row_labels=DActives.index, var_labels=DActives.columns)
    acm.fit(DActives.values)
    
    # Results formatting
    eig = pd.DataFrame(acm.eig_).transpose()
    eig.columns = ['Eigenvalue', 'Percentage', 'Cumulative Percentage']
    
    print("\n### Success! Eigenvalues (Head): ###")
    print(eig.head())

except Exception as e:
    print(f"\n### Error still occurring: ###")
    print(e)
    # Si ça plante encore, on regarde précisément quelle colonne pose problème
    for c in DActives.columns:
        print(f"Value counts for {c}:")
        print(DActives[c].value_counts())
        print("-" * 20)
```

## Une possibilité d'interprétation: distance des individus par rapport au profil moyen

```python
#Profil individu moyen
ind_moy = np.sum(X.values,axis=0)/(n*p)
print(ind_moy)
```

```python
### Ajouter une colonne avec la distance chi-2 de chaque individu par rapport à l'individu moyen
# pour chaque individu: les individus plus éloignés sont plus rares
X['dist_org'] = X.apply(lambda x: round(np.sum(1/ind_moy*(x/p-ind_moy)**2),4), raw=True, axis=1)
X['dist_org']
```

```python
### Distribution des distances à l'individu moyen

sns.set_theme(style="whitegrid",rc={"figure.figsize":(12,2)} )


a = X['dist_org']

print(a.describe())

# ax = sns.boxplot(x=a)
ax = sns.violinplot(x=a)

### Noter que au delà des limites les valeurs sont coupées car postulées
ax.set_xlim(left=min(a), right=max(a))

plt.show()
```

```python
### Individus proches du profil moyen
#  donc fréquents
i = X[(X.dist_org<4.2) & (X.dist_org>3.8)]
print(len(i))
df_pm.loc[i.index][15:20]
```

```python
### Individus moyennements distants du profil moyen
sel = X[(X.dist_org>4.2) & (X.dist_org <6.5)]   ### [X.dist_org<2.5] proches du prof. m.
print(len(sel))
df_pm.loc[sel.index][15:20]
```
